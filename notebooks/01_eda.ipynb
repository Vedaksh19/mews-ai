```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import sys
from pathlib import Path

# Add src to sys.path
sys.path.append(str(Path('../src')))
from eda import run_eda
```

```python
df = pd.read_csv('../data/mews_dataset_v2.csv')
print('Dataset Shape:', df.shape)
df.head()
```

```python
print('--- Descriptive Statistics ---')
df.describe().T
```

```python
print('--- Missing Value Percentages ---')
(df.isna().mean() * 100).round(2)
```

```python
print('--- Class Balance ---')
print(df['manipulation_label'].value_counts(normalize=True).round(4))

print('\n--- Manipulation Rate by Market Cap Class ---')
print(df.groupby('market_cap_class')[['manipulation_label', 'risk_score']].mean().round(4))
```

```python
num_cols = [
    'close_price', 'volume', 'volume_ratio_20d', 'volatility',
    'delivery_pct', 'order_book_imbalance', 'sentiment_score',
    'social_mentions', 'avg_trade_size', 'past_manipulation_flags'
]
print('--- Skewness ---')
df[num_cols].skew().round(2)
```

```python
# Welch t-test on volume_ratio_20d
a = df[df['manipulation_label'] == 1]['volume_ratio_20d'].dropna()
b = df[df['manipulation_label'] == 0]['volume_ratio_20d'].dropna()
t_stat, p_val = stats.ttest_ind(a, b, equal_var=False)
print(f'Welch t-test: t = {t_stat:.2f}, p-value = {p_val:.2e}')
print(f'Mean manipulation volume ratio: {a.mean():.2f} vs Normal: {b.mean():.2f}')
```
